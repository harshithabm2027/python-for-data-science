📝 1. Introduction to OOP

    Object-Oriented Programming (OOP) is a method of structuring programs using classes and objects to represent real-world entities.
    

🧩 2. Basic Concepts
   
   | Concept                      | Description                    | Example                       |
   | ---------------------------- | ------------------------------ | ---------------------------   |
   | Class                        | Blueprint for creating objects | `class Teacher:`               |
   | Object                       | Instance of a class            | `t1 = Teacher()`              |
   | Constructor (`__init__`)     | Initializes object             | `def __init__(self, subject):` |
   | Methods                      | Functions inside class         | `def display(self):`        |
   | Attributes                   | Variables inside class         | `self.subject = subject        |


💻 3. Example Code

```python
# Example 1: Creating a simple class
class Teacher:
    def __init__(self, subject, marks):
        self.subject = subject
        self.marks = marks

    def display(self):
        print(f"Subject: {self.subject}, Marks: {self.marks}")

# Creating objects
t1 = Teacher("Science", 21)
t2 = Teacher("Maths", 43)

t1.display()
t2.display()
```

🧱 4. OOPs Core Concepts with Examples
🔸 Encapsulation
Hiding data inside class using private variables.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def get_balance(self):
        return self.balance

acc = Account(5000)
acc.deposit(2000)
print(acc.get_balance())
```

🔸 Inheritance
Using properties of one class in another.

```python
class Animal:
    def sound(self):
        print("Animals make sounds")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

🔸 Polymorphism
Same method name, different behavior.

```python
class Bird:
    def fly(self):
        print("Birds can fly")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly")

for bird in [Bird(), Penguin()]:
    bird.fly()
```

🔸 Abstraction
   Using abstract classes (via abc module).

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started!")

v = Car()
v.start()
```

🧩 Mini Project :

Library Management System


```python
# Mini Project: Library Management System

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def __str__(self):
        status = "Available" if self.is_available else "Issued"
        return f"{self.title} by {self.author} — {status}"


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(f"✅ '{book.title}' added to {self.name} library.")

    def display_books(self):
        print(f"\n Books in {self.name} Library:")
        for book in self.books:
            print(f" - {book}")

    def issue_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower() and book.is_available:
                book.is_available = False
                print(f"📖 You have issued '{book.title}'. Enjoy reading!")
                return
        print(f" Sorry, '{title}' is not available right now.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower() and not book.is_available:
                book.is_available = True
                print(f" '{book.title}' has been returned. Thank you!")
                return
        print(f"'{title}' was not borrowed from this library.")
library = Library("CMRIT Central Library")

# Add books
library.add_book(Book("Python for Data Science", "Jake VanderPlas"))
library.add_book(Book("Machine Learning Basics", "Andrew Ng"))
library.add_book(Book("Deep Learning with Python", "François Chollet"))

# Display all books
library.display_books()

# Issue a book
library.issue_book("Python for Data Science")

# Try issuing same book again
library.issue_book("Python for Data Science")

# Return the book
library.return_book("Python for Data Science")

# Display updated list
library.display_books()
```

Mini Project : Library Management System 📚

🔍 Explanation
Book Class     : Defines a book with title, author, and availability status.  
    
Library Class  : Contains methods to add, display, issue, and return books. 
    
Demonstrates   : Encapsulation, Objects, and Methods concepts in OOPs.
